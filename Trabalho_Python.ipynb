20. Matriz Transposta

```python
## 20
matriz = [
    [1, 2, 3],
    [4, 5, 6]
]

transposta = []  ## lista vazia

for coluna in range(len(matriz[0])):  ## matriz[0] = primeira linha, o len fala quantos elementos tem nessa linha e o range conta criando a nova linha
    nova_linha = []

    for linha in range(len(matriz)): ## mesma coisa de cima so que pra linha
        nova_linha.append(matriz[linha][coluna])

    transposta.append(nova_linha)  ## .append serve pra colocar numero dentro da nova lista

print("Matriz transposta:")
for linha in transposta:
    print(linha)
```

21.Busca Binária


o Implemente a busca binária para encontrar um elemento em uma
lista ordenada

```python
## 21
lista = [1, 3, 5, 7, 9, 11, 13, 15] ## busca binária so serve pra tabelas organizadas

numero = int(input("Digite o número que deseja encontrar: "))

inicio = 0
fim = len(lista) - 1

encontrado = False

while inicio <= fim:
    meio = (inicio + fim) // 2

    if lista[meio] == numero:
        encontrado = True
        break

    elif lista[meio] < numero:
        inicio = meio + 1

    else:
        fim = meio - 1


if encontrado:
    print("Número encontrado!")
else:
    print("Número não encontrado!")
```

22. Criptografia de Cifra de César


o Crie uma função para criptografar uma mensagem usando a cifra de
César.


```python
## 22
def criptografar(mensagem, deslocamento): ## cria função criptografar que vai receber esses 2 valores
    resultado = ""  ## vai guardar o resultado da mensagem

    for letra in mensagem:  #vai "desmembrar" a mensagem ex: casa = c, a, s, a
        if letra.isalpha(): ## isalpha verifica se é uma letra
            nova_letra = chr(ord(letra) + deslocamento) ## ord transforma a letra em um número ex: a = 5 então c = 7 chr faz o contrario
            resultado += nova_letra
        else:
            resultado += letra

    return resultado


mensagem = input("Digite uma mensagem: ")
chave = int(input("Digite o deslocamento: "))

criptografada = criptografar(mensagem, chave)

print("Mensagem criptografada:", criptografada)
```

23. Descriptografar Cifra de César


o Crie uma função para descriptografar uma mensagem cifrada com a
cifra de César.


```python
## 23
def descriptografar(mensagem, deslocamento):
    resultado = ""

    for letra in mensagem:
        if letra.isalpha():
            nova_letra = chr(ord(letra) - deslocamento) ## a unica diferença pro codigo anterior é a gente tirar o deslocamento em vez de adicionar
            resultado += nova_letra
        else:
            resultado += letra

    return resultado


mensagem = input("Digite a mensagem criptografada: ")
chave = int(input("Digite a chave: "))

texto = descriptografar(mensagem, chave)

print("Mensagem original:", texto)
```

24.Validação de CPF


o Implemente uma função para validar um CPF brasileiro.

```python
## 24
def validar_cpf(cpf):
    cpf = cpf.replace(".", "").replace("-", "") ## replace é uma função que vai trocar alguma coisa por outra, nesse caso . e - por nada pro cpf ser formatado sem pontos ou traços

    if len(cpf) != 11 or not cpf.isdigit(): ## verifica se o cpf tem 11 digitos com o len, e se todos caracteres digitados são números
        return False

    soma = 0
    peso = 10
## Os cpfs tem uma regra para saber se é valido nesse caso
## iremos calcular os 9 primeiros numeros do cpf pelo peso
## exemplo: cpf: 123456789 calculo: 1 x 10, 2 x 9, 3 x 8 ...
    for n in range(9):
        soma += int(cpf[n]) * peso
        peso -= 1

    resto = (soma * 10) % 11 ## após descobrirmos a soma usamos essa fórmula
    ## para descobrirmos o digito verificador

    if resto == 10: ## caso o resultado da formula seja 10, o numero vira 0
        resto = 0

    if resto != int(cpf[9]):
        return False

    soma = 0
    peso = 11
## aqui começa a 2° parte da fórmula onde iremos fazer o mesmo cálculo só que
## implementando o 1° digito verificador e o peso começando em 11 em vez de 10
    for n in range(10):
        soma += int(cpf[n]) * peso
        peso -= 1

    resto = (soma * 10) % 11

    if resto == 10:
        resto = 0

    if resto != int(cpf[10]):
        return False

    return True
##resumidamente se os 9 primeiros números corresponderem aos ultimos 2 digitos o cpf vai ser validado

cpf = input("Digite o CPF: ")

if validar_cpf(cpf):
    print("CPF válido")
else:
    print("CPF inválido")
```

30.Jogo da Forca


o Desenvolva um jogo simples da forca onde o usuário deve adivinhar
uma palavra.


```python
## 30
palavra = "nychollas"

tentativas = 6
letras_acertadas = []

for letra in palavra:
    letras_acertadas.append("_") ## conta quantas letras tem na palavra e substitui por espaços vazios

while tentativas > 0: ## enquanto tentativas for maior que 0 continue jogando
    print("Palavra:", " ".join(letras_acertadas)) ## .join é usado para juntar lista em um unica string sem virgulas e espaços
    print("Tentativas restantes:", tentativas)

    tentativa = input("Digite uma letra: ").lower() ## . lower formata em mínuscula

    if tentativa in palavra:
        print("Você acertou!")

        for n in range(len(palavra)):
            if palavra[n] == tentativa:
                letras_acertadas[n] = tentativa ## o programa nomeia cada letra com um número e testa sua letra em todas as posições se vc acertar substitui o vazio pela letra que vc tentou

    else:
        print("Você errou!")
        tentativas -= 1 ## se errar perde uma tentativa

    if "_" not in letras_acertadas: ## not in serve pra ver se ainda existe algum espaço vazio, se estiver tudo preenchido vc acertou a palavra
        print("\nParabéns! Você descobriu a palavra:", palavra)
        break

if tentativas == 0:
    print("\nVocê perdeu! A palavra era:", palavra)
```

31.Ordenação por Bolha


o Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
## 31
numeros = [5, 3, 8, 4, 2]
for i in range(len(numeros) - 1): ## len ve que nossa lista tem 5 numeros,  entao fica range(5-1) que é 0,1,2,3

    for n in range(len(numeros) - 1 - i): ## usamos n e i para o código funcionar em listas de qualquer tamanho, o -1 porque se acharmos 4 numeros certos o 5° automaticamente tb vai estar certo

        #se o número da esquerda é maior que o da direita
        if numeros[n] > numeros[n + 1]:

            # o código troca
            numeros[n], numeros[n + 1] = numeros[n + 1], numeros[n]

# Mostra o resultado
print("numeros ordenado:{}".format(numeros))
```

37. Funções Recursivas


o Implemente uma função recursiva para calcular o fatorial de um
número.


```python
def fat(n): ##cria uma função chamada fat
    if n == 0: ## 0! é igual a 1
        return 1
    else:
        return n * fat(n - 1) ## realizando a função recursiva que chama ela mesma até ser 0


numero = int(input("Digite um número: "))

resultado = fat(numero)

print("O fatorial é:", resultado)

## caso a pessoa digite 0, sabemos que o resultado é 1, se não a função faz a conta e chama ela mesma varias vezes até não precisar multiplicar mais
```

```python
lista = [] ## criando a lista

while True: ## enquanto essas condiçoes forem verdadeiras continue
    print("\n1 - Adicionar item")
    print("2 - Remover item")
    print("3 - Listar itens")
    print("4 - Sair")

    opcao = input("Escolha uma opção: ")  ## escolher uma opção

    if opcao == "1":
        item = input("Digite o item que deseja adicionar: ")
        lista.append(item)  ##fatorar a palavra para nao dar erro
        print(item, "foi adicionado.")

    elif opcao == "2":
        item = input("Digite o item que deseja remover: ")

        if item in lista:
            lista.remove(item)  ## remover da lista
            print(item, "foi removido.")
        else:
            print("Esse item não está na lista.")

    elif opcao == "3":
        print("Lista de compras:")

        for item in lista:
            print("-", item)  ## listar toda a lista usando o for fazendo percorrer toda a lista

    elif opcao == "4":
        print("Programa encerrado.")
        break ## encerrar

    else:
        print("Opção inválida.")

        ## if elifs e else junto com alguns comandos de lista para excluir, for para listar, else em qualquer caso de erro e while para voltar ao começo
```

41.Remover Duplicatas

o Remova duplicatas de uma lista de números.


```python
numeros = [1, 2, 2, 3, 4, 4, 5] ## lista de numeros

nsd = [] ## lista sem duplicatas

for numero in numeros:
    if numero not in nsd: ## se o numero nao estiver na lista sem duplicatas, adiciona
        nsd.append(numero) ## adiciona o numero na lista

print("Lista sem duplicatas:", nsd)

## se o numero da lista 1 ja estiver na lista 2 nao adiciona novamente, removendo as duplicatas
```

47. Contar Números em Lista


o Conte quantas vezes cada número aparece em uma lista.

```python
numeros = [1, 2, 2, 3, 3, 3]

contagem = {} ## {} cria um dicionario ordenando os numeros em colunas

for numero in numeros:
    if numero in contagem: ## for percorre a lista, se o numero nao foi contado soma + 1
        contagem[numero] = contagem[numero] + 1
    else:
        contagem[numero] = 1 ## se ja foi permanece o mesmo

print(contagem)

## usamos o {} para guardar os numeros que foram sendo contados e if e else para ordenar quais foram e quais nao foram
```

48.Ordenação por Inserção


o Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
numeros = [5, 3, 4, 1, 2]

for i in range(1, len(numeros)): ## len conta a quantidade
    atual = numeros[i] ## guarda o numero que queremos ordenar
    j = i - 1 ## pega a posição anterior

    while j >= 0 and numeros[j] > atual: ## enquanto j for maior que 0 e o numero anterior for maior que o atual
        numeros[j + 1] = numeros[j] ## move o numero para direita
        j = j - 1 ## volta uma posição para continuar procurando onde ordenar os numeros

    numeros[j + 1] = atual ## coloca o numero atual na posiçao correta

print(numeros)

## usamos o while para sempre procurar onde colocar os numeros ate o numero anterior nao ser maior que o atual, o que sifnigica que esta ordenado
```

50. Cálculo de Frequência de Letras


o Calcule a frequência de cada letra em uma string fornecida pelo
usuário.

```python
texto = input("Digite um texto: ").lower() ##.lower transforma tudo em minusculo

frequencia = {} ## cria um dicionario vazio

for letra in texto: ## percorre letra por letra
    if letra in frequencia:
        frequencia[letra] = frequencia[letra] + 1 ## se a letra ja apareceu antes soma 1
    else:
        frequencia[letra] = 1 ## se nao apareceu permanece o mesmo

print(frequencia)

## .lower formata tudo em minusculo, for percorre as letras, se a letra ja apareceu antes soma 1, se nao permanece
```
